```python
# ============================================================
# COMPLETE DATA ANALYSIS
# ============================================================

import pandas as pd
import numpy as np

# ============================================================
# 1. LOAD CLEANED DATASET
# ============================================================

file_path = "/content/cleaned_sample_dataset.xlsx"

df = pd.read_excel(file_path)

print("=" * 60)
print("TASK 19 - ENTERPRISE ANALYTICS")
print("=" * 60)

print("\nDataset Shape:", df.shape)

print("\nFirst 5 Records:")
display(df.head())


# ============================================================
# 2. BASIC DATASET INFORMATION
# ============================================================

print("\n" + "=" * 60)
print("1. DATASET INFORMATION")
print("=" * 60)

print("Total Records :", len(df))
print("Total Columns :", len(df.columns))

if "tenant_id" in df.columns:
    print("Total Tenants :", df["tenant_id"].nunique())

if "industry" in df.columns:
    print("Total Industries :", df["industry"].nunique())

if "location" in df.columns:
    print("Total Locations :", df["location"].nunique())


# ============================================================
# 3. DATA QUALITY CHECK
# ============================================================

print("\n" + "=" * 60)
print("2. DATA QUALITY ANALYSIS")
print("=" * 60)

print("Duplicate Records :", df.duplicated().sum())

missing = df.isnull().sum()

quality_report = pd.DataFrame({
    "Column": missing.index,
    "Missing_Values": missing.values,
    "Missing_Percentage":
        (missing.values / len(df) * 100).round(2)
})

print("\nMissing Value Report:")
display(
    quality_report[
        quality_report["Missing_Values"] > 0
    ]
)


# ============================================================
# 4. OVERALL KPI ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("3. OVERALL KPI ANALYSIS")
print("=" * 60)

total_applications = df["applications"].sum()
total_screened = df["screened"].sum()
total_interviewed = df["interviewed"].sum()
total_offered = df["offered"].sum()
total_hired = df["hired"].sum()

total_cost = df["hiring_cost"].sum()
total_revenue = df["revenue_value"].sum()

average_quality = df["quality_score"].mean()
average_time_to_hire = df["time_to_hire_days"].mean()

overall_roi = (
    (total_revenue - total_cost)
    / total_cost
) * 100


print("Total Applications :", total_applications)
print("Total Screened     :", total_screened)
print("Total Interviewed  :", total_interviewed)
print("Total Offers       :", total_offered)
print("Total Hires        :", total_hired)

print("\nTotal Hiring Cost   :", round(total_cost, 2))
print("Total Revenue Value:", round(total_revenue, 2))

print("\nAverage Quality Score:",
      round(average_quality, 2))

print("Average Time to Hire:",
      round(average_time_to_hire, 2),
      "days")

print("Overall ROI:",
      round(overall_roi, 2), "%")


# ============================================================
# 5. FUNNEL ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("4. RECRUITMENT FUNNEL ANALYSIS")
print("=" * 60)

screening_rate = (
    total_screened / total_applications * 100
)

interview_rate = (
    total_interviewed / total_screened * 100
)

offer_rate = (
    total_offered / total_interviewed * 100
)

hire_rate = (
    total_hired / total_offered * 100
)

overall_hire_rate = (
    total_hired / total_applications * 100
)


funnel_analysis = pd.DataFrame({
    "Stage": [
        "Applications",
        "Screened",
        "Interviewed",
        "Offered",
        "Hired"
    ],
    "Count": [
        total_applications,
        total_screened,
        total_interviewed,
        total_offered,
        total_hired
    ]
})

funnel_analysis["Conversion_Rate_Percent"] = [
    100,
    round(screening_rate, 2),
    round(interview_rate, 2),
    round(offer_rate, 2),
    round(hire_rate, 2)
]

display(funnel_analysis)

print(
    "\nOverall Application-to-Hire Rate:",
    round(overall_hire_rate, 2),
    "%"
)


# ============================================================
# 6. INDUSTRY ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("5. INDUSTRY PERFORMANCE ANALYSIS")
print("=" * 60)

industry_analysis = (
    df.groupby("industry")
    .agg(
        Total_Applications=("applications", "sum"),
        Total_Hired=("hired", "sum"),
        Average_Quality=("quality_score", "mean"),
        Average_Time_to_Hire=("time_to_hire_days", "mean"),
        Total_Cost=("hiring_cost", "sum"),
        Total_Revenue=("revenue_value", "sum")
    )
    .reset_index()
)

industry_analysis["Hire_Rate_Percent"] = (
    industry_analysis["Total_Hired"]
    / industry_analysis["Total_Applications"]
    * 100
)

industry_analysis["ROI_Percent"] = (
    (
        industry_analysis["Total_Revenue"]
        - industry_analysis["Total_Cost"]
    )
    / industry_analysis["Total_Cost"]
    * 100
)

industry_analysis = industry_analysis.round(2)

display(industry_analysis)


# ============================================================
# 7. TENANT PERFORMANCE ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("6. TENANT PERFORMANCE ANALYSIS")
print("=" * 60)

tenant_analysis = (
    df.groupby("tenant_id")
    .agg(
        Applications=("applications", "sum"),
        Screened=("screened", "sum"),
        Interviewed=("interviewed", "sum"),
        Offers=("offered", "sum"),
        Hired=("hired", "sum"),
        Avg_Quality=("quality_score", "mean"),
        Avg_Time_to_Hire=("time_to_hire_days", "mean"),
        Total_Cost=("hiring_cost", "sum"),
        Total_Revenue=("revenue_value", "sum")
    )
    .reset_index()
)

tenant_analysis["Hire_Rate_Percent"] = (
    tenant_analysis["Hired"]
    / tenant_analysis["Applications"]
    * 100
)

tenant_analysis["ROI_Percent"] = (
    (
        tenant_analysis["Total_Revenue"]
        - tenant_analysis["Total_Cost"]
    )
    / tenant_analysis["Total_Cost"]
    * 100
)

tenant_analysis = tenant_analysis.round(2)

display(tenant_analysis)


# ============================================================
# 8. TOP & BOTTOM TENANTS
# ============================================================

print("\n" + "=" * 60)
print("7. TOP PERFORMING TENANTS")
print("=" * 60)

top_tenants = tenant_analysis.sort_values(
    "ROI_Percent",
    ascending=False
).head(5)

display(top_tenants)


print("\n" + "=" * 60)
print("8. LOW PERFORMING TENANTS")
print("=" * 60)

bottom_tenants = tenant_analysis.sort_values(
    "ROI_Percent",
    ascending=True
).head(5)

display(bottom_tenants)


# ============================================================
# 9. QUALITY ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("9. QUALITY ANALYSIS")
print("=" * 60)

quality_analysis = (
    df.groupby("quality_category")
    .agg(
        Records=("candidate_id", "count"),
        Applications=("applications", "sum"),
        Hired=("hired", "sum"),
        Average_Quality=("quality_score", "mean"),
        Average_Time_to_Hire=("time_to_hire_days", "mean")
    )
    .reset_index()
)

quality_analysis["Hire_Rate_Percent"] = (
    quality_analysis["Hired"]
    / quality_analysis["Applications"]
    * 100
)

quality_analysis = quality_analysis.round(2)

display(quality_analysis)


# ============================================================
# 10. TIME-TO-HIRE ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("10. TIME-TO-HIRE ANALYSIS")
print("=" * 60)

time_analysis = (
    df.groupby("time_to_hire_category")
    .agg(
        Records=("candidate_id", "count"),
        Hired=("hired", "sum"),
        Average_Time=("time_to_hire_days", "mean"),
        Average_Quality=("quality_score", "mean"),
        Total_Revenue=("revenue_value", "sum")
    )
    .reset_index()
)

display(time_analysis.round(2))


print("\nFastest Average Hiring Time:",
      round(df["time_to_hire_days"].min(), 2),
      "days")

print("Slowest Average Hiring Time:",
      round(df["time_to_hire_days"].max(), 2),
      "days")


# ============================================================
# 11. CONFIGURATION ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("11. CONFIGURATION IMPACT ANALYSIS")
print("=" * 60)


# Auto Screening
auto_screening_analysis = (
    df.groupby("auto_screening")
    .agg(
        Records=("candidate_id", "count"),
        Applications=("applications", "sum"),
        Hired=("hired", "sum"),
        Avg_Quality=("quality_score", "mean"),
        Avg_Time_to_Hire=("time_to_hire_days", "mean"),
        Total_Cost=("hiring_cost", "sum"),
        Total_Revenue=("revenue_value", "sum")
    )
    .reset_index()
)

auto_screening_analysis["Hire_Rate_Percent"] = (
    auto_screening_analysis["Hired"]
    / auto_screening_analysis["Applications"]
    * 100
)

auto_screening_analysis["ROI_Percent"] = (
    (
        auto_screening_analysis["Total_Revenue"]
        - auto_screening_analysis["Total_Cost"]
    )
    / auto_screening_analysis["Total_Cost"]
    * 100
)

print("\nAuto Screening Analysis:")
display(auto_screening_analysis.round(2))


# ============================================================
# 12. INTERVIEW STAGE ANALYSIS
# ============================================================

interview_stage_analysis = (
    df.groupby("interview_stages")
    .agg(
        Records=("candidate_id", "count"),
        Applications=("applications", "sum"),
        Hired=("hired", "sum"),
        Avg_Quality=("quality_score", "mean"),
        Avg_Time_to_Hire=("time_to_hire_days", "mean"),
        Total_Cost=("hiring_cost", "sum"),
        Total_Revenue=("revenue_value", "sum")
    )
    .reset_index()
)

interview_stage_analysis["Hire_Rate_Percent"] = (
    interview_stage_analysis["Hired"]
    / interview_stage_analysis["Applications"]
    * 100
)

interview_stage_analysis["ROI_Percent"] = (
    (
        interview_stage_analysis["Total_Revenue"]
        - interview_stage_analysis["Total_Cost"]
    )
    / interview_stage_analysis["Total_Cost"]
    * 100
)

print("\nInterview Stage Analysis:")
display(interview_stage_analysis.round(2))


# ============================================================
# 13. APPROVAL WORKFLOW ANALYSIS
# ============================================================

workflow_analysis = (
    df.groupby("approval_workflow")
    .agg(
        Records=("candidate_id", "count"),
        Applications=("applications", "sum"),
        Hired=("hired", "sum"),
        Avg_Quality=("quality_score", "mean"),
        Avg_Time_to_Hire=("time_to_hire_days", "mean"),
        Total_Cost=("hiring_cost", "sum"),
        Total_Revenue=("revenue_value", "sum")
    )
    .reset_index()
)

workflow_analysis["Hire_Rate_Percent"] = (
    workflow_analysis["Hired"]
    / workflow_analysis["Applications"]
    * 100
)

workflow_analysis["ROI_Percent"] = (
    (
        workflow_analysis["Total_Revenue"]
        - workflow_analysis["Total_Cost"]
    )
    / workflow_analysis["Total_Cost"]
    * 100
)

print("\nApproval Workflow Analysis:")
display(workflow_analysis.round(2))


# ============================================================
# 14. BENCHMARK ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("12. BENCHMARK ANALYSIS")
print("=" * 60)

benchmark = {
    "Average Quality Score":
        df["quality_score"].mean(),

    "Average Time to Hire":
        df["time_to_hire_days"].mean(),

    "Overall Hire Rate":
        overall_hire_rate,

    "Overall ROI":
        overall_roi
}

benchmark_df = pd.DataFrame(
    benchmark.items(),
    columns=["Metric", "Benchmark_Value"]
)

benchmark_df["Benchmark_Value"] = (
    benchmark_df["Benchmark_Value"].round(2)
)

display(benchmark_df)


# ============================================================
# 15. TENANT BENCHMARK COMPARISON
# ============================================================

tenant_analysis["Quality_vs_Benchmark"] = (
    tenant_analysis["Avg_Quality"]
    - df["quality_score"].mean()
)

tenant_analysis["Time_vs_Benchmark"] = (
    tenant_analysis["Avg_Time_to_Hire"]
    - df["time_to_hire_days"].mean()
)

tenant_analysis["ROI_vs_Benchmark"] = (
    tenant_analysis["ROI_Percent"]
    - overall_roi
)

print("\n" + "=" * 60)
print("13. TENANT BENCHMARK COMPARISON")
print("=" * 60)

display(
    tenant_analysis[
        [
            "tenant_id",
            "Avg_Quality",
            "Avg_Time_to_Hire",
            "ROI_Percent",
            "Quality_vs_Benchmark",
            "Time_vs_Benchmark",
            "ROI_vs_Benchmark"
        ]
    ].round(2)
)


# ============================================================
# 16. RECOMMENDATION FLAGS
# ============================================================

def recommendation(row):

    recommendations = []

    if row["ROI_Percent"] < overall_roi:
        recommendations.append("Improve ROI")

    if row["Avg_Quality"] < average_quality:
        recommendations.append("Improve Candidate Quality")

    if row["Avg_Time_to_Hire"] > average_time_to_hire:
        recommendations.append("Reduce Time-to-Hire")

    if row["Hire_Rate_Percent"] < overall_hire_rate:
        recommendations.append("Improve Funnel Conversion")

    if len(recommendations) == 0:
        return "Performing Above Benchmark"

    return "; ".join(recommendations)


tenant_analysis["Recommendation"] = (
    tenant_analysis.apply(
        recommendation,
        axis=1
    )
)


print("\n" + "=" * 60)
print("14. TENANT RECOMMENDATIONS")
print("=" * 60)

display(
    tenant_analysis[
        [
            "tenant_id",
            "Hire_Rate_Percent",
            "Avg_Quality",
            "Avg_Time_to_Hire",
            "ROI_Percent",
            "Recommendation"
        ]
    ].round(2)
)


# ============================================================
# 17. FINAL MANAGEMENT SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("15. MANAGEMENT SUMMARY")
print("=" * 60)

best_roi_tenant = tenant_analysis.loc[
    tenant_analysis["ROI_Percent"].idxmax()
]

worst_roi_tenant = tenant_analysis.loc[
    tenant_analysis["ROI_Percent"].idxmin()
]

best_quality_tenant = tenant_analysis.loc[
    tenant_analysis["Avg_Quality"].idxmax()
]

fastest_tenant = tenant_analysis.loc[
    tenant_analysis["Avg_Time_to_Hire"].idxmin()
]


print("\nOverall Business Performance")
print("--------------------------------")
print("Applications :", total_applications)
print("Hires        :", total_hired)
print("Hire Rate    :", round(overall_hire_rate, 2), "%")
print("Avg Quality  :", round(average_quality, 2))
print("Avg Time     :", round(average_time_to_hire, 2), "days")
print("Total Cost   :", round(total_cost, 2))
print("Total Revenue:", round(total_revenue, 2))
print("Overall ROI  :", round(overall_roi, 2), "%")


print("\nBest ROI Tenant:")
print(best_roi_tenant[
    ["tenant_id", "ROI_Percent"]
].to_dict())


print("\nLowest ROI Tenant:")
print(worst_roi_tenant[
    ["tenant_id", "ROI_Percent"]
].to_dict())


print("\nHighest Quality Tenant:")
print(best_quality_tenant[
    ["tenant_id", "Avg_Quality"]
].to_dict())


print("\nFastest Hiring Tenant:")
print(fastest_tenant[
    ["tenant_id", "Avg_Time_to_Hire"]
].to_dict())


# ============================================================
# 18. EXPORT ANALYSIS RESULTS
# ============================================================

output_file = "/content/task19_analysis_results.xlsx"

with pd.ExcelWriter(output_file, engine="openpyxl") as writer:

    df.to_excel(
        writer,
        sheet_name="Cleaned_Data",
        index=False
    )

    funnel_analysis.to_excel(
        writer,
        sheet_name="Funnel_Analysis",
        index=False
    )

    industry_analysis.to_excel(
        writer,
        sheet_name="Industry_Analysis",
        index=False
    )

    tenant_analysis.to_excel(
        writer,
        sheet_name="Tenant_Analysis",
        index=False
    )

    quality_analysis.to_excel(
        writer,
        sheet_name="Quality_Analysis",
        index=False
    )

    time_analysis.to_excel(
        writer,
        sheet_name="Time_to_Hire",
        index=False
    )

    auto_screening_analysis.to_excel(
        writer,
        sheet_name="Auto_Screening",
        index=False
    )

    interview_stage_analysis.to_excel(
        writer,
        sheet_name="Interview_Stages",
        index=False
    )

    workflow_analysis.to_excel(
        writer,
        sheet_name="Approval_Workflow",
        index=False
    )

    benchmark_df.to_excel(
        writer,
        sheet_name="Benchmarks",
        index=False
    )

print("\n" + "=" * 60)
print("ANALYSIS COMPLETED SUCCESSFULLY")
print("=" * 60)

print("\nFinal Analysis File:")
print(output_file)

print("\nUse this Excel file for Tableau:")
print("task19_analysis_results.xlsx")
```
